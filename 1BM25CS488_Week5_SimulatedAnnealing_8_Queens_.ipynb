```python

import random
import math


def calculate_cost(state):
    """Count pairs of queens attacking each other."""
    conflicts = 0
    n = len(state)

    for i in range(n):
        for j in range(i + 1, n):
            # Same row or same diagonal
            if state[i] == state[j] or abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1

    return conflicts


def random_neighbor(state):
    """Generate a random neighbor by swapping two queens."""
    neighbor = state.copy()

    i, j = random.sample(range(len(state)), 2)
    neighbor[i], neighbor[j] = neighbor[j], neighbor[i]

    return neighbor


def simulated_annealing(initial_state):
    current = initial_state.copy()
    temperature = 100.0
    cooling_rate = 0.99
    minimum_temperature = 0.1
    step = 0

    while temperature > minimum_temperature:
        current_cost = calculate_cost(current)

        print(f"\nStep {step}")
        print(f"Current state: {current}")
        print(f"Current cost: {current_cost}")
        print(f"Temperature: {temperature:.4f}")

        # Goal test
        if current_cost == 0:
            print("\nGoal configuration reached!")
            return current

        # Generate a random neighbor
        next_state = random_neighbor(current)
        next_cost = calculate_cost(next_state)

        # Delta E = current cost - next cost
        delta_e = current_cost - next_cost

        print(f"Neighbor state: {next_state}")
        print(f"Neighbor cost: {next_cost}")
        print(f"Delta E: {delta_e}")

        if delta_e > 0:
            # Better state: accept it
            current = next_state
            print("Better state: Accepted")
        else:
            # Equal or worse state: accept probabilistically
            probability = math.exp(delta_e / temperature)
            random_number = random.random()

            print(f"Acceptance probability: {probability:.4f}")
            print(f"Random number: {random_number:.4f}")

            if random_number < probability:
                current = next_state
                print("Accepted probabilistically")
            else:
                print("Neighbor rejected")

        # Decrease temperature
        temperature *= cooling_rate
        step += 1

    # Check the final state
    if calculate_cost(current) == 0:
        print("\nGoal configuration reached!")
    else:
        print("\nTemperature reached the stopping limit.")
        print("No solution found in this run.")

    return current


def display_board(state):
    """Display the chessboard."""
    n = len(state)

    print("\nChessboard:")
    for row in range(n):
        for col in range(n):
            if state[col] == row:
                print(" Q ", end="")
            else:
                print(" . ", end="")
        print()


# Main program
n = 8

print("8-Queens Problem using Simulated Annealing")
print("Enter 8 row positions from 0 to 7.")
print("Example: 0 4 7 5 2 6 1 3")

initial_state = list(map(
    int, input("Enter initial state: ").split()
))

if (len(initial_state) != n
        or sorted(initial_state) != list(range(n))):
    print("Invalid input! Enter each number from 0 to 7 exactly once.")
else:
    final_state = simulated_annealing(initial_state)

    print("\nFinal state:", final_state)
    print("Final cost:", calculate_cost(final_state))

    if calculate_cost(final_state) == 0:
        display_board(final_state)
```
