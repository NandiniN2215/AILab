```python

print("Nandini N")
print("1BM25CS488")


def forward_chaining(facts, rules, query):

    facts = set(facts)
    iteration = 1

    print("\n========== FORWARD CHAINING ==========")

    print("\nInitially known facts:")

    for fact in facts:
        print("-", fact)

    while True:

        new = []

        for premises, conclusion in rules:

            if all(p in facts for p in premises):

                if conclusion not in facts:
                    new.append((premises, conclusion))

        if not new:
            break

        print("\nIteration", iteration, ":")

        for premises, conclusion in new:

            print("\nSince")

            for p in premises:
                print("-", p)

            print("We can conclude that:")
            print(conclusion)

            facts.add(conclusion)

        iteration += 1

        if query in facts:
            return True

    return query in facts


# Enter facts

n = int(input("\nEnter number of facts: "))

facts = []

for i in range(n):

    fact = input("Enter fact: ").strip()
    facts.append(fact)


# Enter rules

r = int(input("\nEnter number of rules: "))

rules = []

for i in range(r):

    print("\nRule", i + 1)

    p = int(input("Enter number of premises: "))

    premises = []

    for j in range(p):

        premise = input("Enter premise: ").strip()
        premises.append(premise)

    conclusion = input("Enter conclusion: ").strip()

    rules.append((premises, conclusion))


# Enter query

query = input("\nEnter query to prove: ").strip()


# Execute forward chaining

result = forward_chaining(facts, rules, query)


# Display result

print("\n========== FINAL RESULT ==========")

if result:

    print("The given query is proved.")
    print("Conclusion:", query)

else:

    print("The given query cannot be proved.")
    print("Conclusion:", query)
```
