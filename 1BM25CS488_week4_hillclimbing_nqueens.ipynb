```python
print("Nandini.N")
print("1BM25CS488")
def calculate_cost(state):
    """Count pairs of queens attacking each other."""
    n = len(state)
    conflicts = 0

    for i in range(n):
        for j in range(i + 1, n):
            if abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1

    return conflicts


def get_neighbors(state):
    """Generate neighbors by swapping rows of two queens."""
    neighbors = []
    n = len(state)

    for i in range(n):
        for j in range(i + 1, n):
            neighbor = state.copy()
            neighbor[i], neighbor[j] = neighbor[j], neighbor[i]

            neighbors.append(
                ((i, j), neighbor, calculate_cost(neighbor))
            )

    return neighbors


def hill_climbing_all(initial_state):
    solutions = []
    visited = set()
    stack = [(initial_state.copy(), [])]

    while stack:
        current, path = stack.pop()
        current_tuple = tuple(current)

        if current_tuple in visited:
            continue

        visited.add(current_tuple)
        current_cost = calculate_cost(current)

        print("\n" + "=" * 50)
        print(f"Step {len(path)}")
        print(f"Current state: {current}")
        print(f"Current cost: {current_cost}")

        if current_cost == 0:
            print("Goal configuration reached!")
            solutions.append((current.copy(), path))
            continue

        neighbors = get_neighbors(current)

        print("\nNeighbor costs:")
        for pair, state, cost in neighbors:
            print(f"Swap {pair}: {state} -> Cost = {cost}")

        # Find the minimum cost among all neighbors
        best_cost = min(cost for _, _, cost in neighbors)

        # Standard hill climbing: only accept improvements
        if best_cost >= current_cost:
            print("\nNo better neighbor exists.")
            print("Local minimum reached.")
            continue

        # Explore every neighbor tied for the best cost
        best_neighbors = [
            (pair, state, cost)
            for pair, state, cost in neighbors
            if cost == best_cost
        ]

        # Reverse push order so the smallest pair is explored first
        best_neighbors.sort(key=lambda item: item[0], reverse=True)

        for pair, state, cost in best_neighbors:
            print(f"\nSelected swap: {pair}")
            print(f"Next state: {state}")
            print(f"Next cost: {cost}")
            stack.append((state, path + [(pair, state, cost)]))

    print("\n" + "=" * 50)
    print("ALL GOAL CONFIGURATIONS FOUND:")

    if not solutions:
        print("No goal configuration found from this initial state.")
    else:
        for i, (state, path) in enumerate(solutions, 1):
            print(f"\nSolution {i}: {state}")
            print("Path:")
            print("Initial state:", initial_state)

            for step, (pair, next_state, cost) in enumerate(path, 1):
                print(
                    f"Step {step}: Swap {pair} -> "
                    f"{next_state}, Cost = {cost}"
                )

    print(f"\nTotal goal configurations found: {len(solutions)}")


# Take input from the user
n = int(input("Enter the number of queens (4): "))

if n != 4:
    print("This program is designed for the 4-Queens problem.")
else:
    print("Enter the row position of each queen (0 to 3).")

    initial_state = list(map(
        int, input("Enter 4 row positions: ").split()
    ))

    if (len(initial_state) != 4
            or any(x < 0 or x >= 4 for x in initial_state)
            or len(set(initial_state)) != 4):
        print("Invalid input! Enter a permutation of 0, 1, 2, 3.")
    else:
        hill_climbing_all(initial_state)
```
