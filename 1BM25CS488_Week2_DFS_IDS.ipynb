```python
print("Nandini N")
print("1BM25CS488")
def dls(state, goal, depth, path):
    if state == goal:
        return path

    if depth == 0:
        return None

    zero = state.index(0)
    r, c = divmod(zero, 3)

    moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]

    for dr, dc in moves:
        nr, nc = r + dr, c + dc

        if 0 <= nr < 3 and 0 <= nc < 3:
            new = state[:]
            pos = nr * 3 + nc

            new[zero], new[pos] = new[pos], new[zero]

            if new not in path:
                result = dls(new, goal, depth - 1,
                             path + [new])
                if result:
                    return result

    return None


def iddfs(start, goal):
    for depth in range(32):
        result = dls(start, goal, depth, [start])

        if result:
            print("Solution found at depth:", depth)

            for step in result:
                for i in range(0, 9, 3):
                    print(step[i:i+3])
                print()

            return

    print("No solution found")


start = [1, 2, 3, 0, 4, 6, 7, 5, 8]
goal = [1, 2, 3, 4, 5, 6, 7, 8, 0]

iddfs(start, goal)
```
