```python
print("Nandini.N")
print("1BM25CS488")
def dls(state, goal, depth, path):
    if state == goal:
        return path

    if depth == 0:
        return None

    zero = state.index(0)
    r, c = divmod(zero, 3)

    moves = [
        (-1, 0, "UP"),
        (1, 0, "DOWN"),
        (0, -1, "LEFT"),
        (0, 1, "RIGHT")
    ]

    for dr, dc, direction in moves:
        nr, nc = r + dr, c + dc

        if 0 <= nr < 3 and 0 <= nc < 3:
            new_pos = nr * 3 + nc
            new_state = list(state)

            new_state[zero], new_state[new_pos] = (
                new_state[new_pos], new_state[zero]
            )

            new_state = tuple(new_state)

            if new_state not in [s for s, _ in path]:
                result = dls(
                    new_state, goal, depth - 1,
                    path + [(new_state, direction)]
                )

                if result is not None:
                    return result

    return None


initial = (1, 2, 3, 0, 4, 6, 7, 5, 8)
goal = (1, 2, 3, 4, 5, 6, 7, 8, 0)

for depth in range(50):
    result = dls(initial, goal, depth, [(initial, "START")])

    if result is not None:
        print("Solution found at depth:", depth)

        for state, direction in result:
            print("\nBlank moves:", direction)
            for i in range(0, 9, 3):
                print(state[i:i+3])
        break
```
